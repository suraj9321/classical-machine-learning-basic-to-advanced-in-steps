### Exercise 1: Create arrays
Create:
- `a`: the integers 0 to 19, as a 4×5 matrix
- `b`: 7 evenly spaced numbers from -1 to 1 (inclusive)

```python
a = np.arange(20).reshape(4, 5)
b = np.linspace(-1, 1, 7)
```

```python
# check
assert a.shape == (4, 5) and a[3, 4] == 19, "a should be 0..19 reshaped to (4, 5)"
assert np.allclose(b, [-1, -2/3, -1/3, 0, 1/3, 2/3, 1]), "b should have 7 points from -1 to 1"
print("✅ Exercise 1 passed")
```

### Exercise 2: Slicing
Using `a` from Exercise 1, extract:
- `last_col`: the last column (1-D)
- `inner`: rows 1–2 and columns 1–3 (a 2×3 block)
- `evens`: all even numbers in `a`, using a boolean mask (1-D)

```python
last_col = a[:, -1]
inner = a[1:3, 1:4]
evens = a[a % 2 == 0]
```

```python
# check
assert np.array_equal(last_col, [4, 9, 14, 19])
assert np.array_equal(inner, [[6, 7, 8], [11, 12, 13]])
assert np.array_equal(evens, np.arange(0, 20, 2))
print("✅ Exercise 2 passed")
```

### Exercise 3: Aggregations along an axis
`X` below holds 100 samples with 3 features. Compute:
- `col_means`: mean of each feature → shape `(3,)`
- `row_max`: the largest value in each sample → shape `(100,)`
- `best_feature_per_row`: the **index** of the largest feature in each sample → shape `(100,)`

```python
X = rng.normal(loc=[10, 0, -5], scale=[1, 2, 3], size=(100, 3))

col_means = X.mean(axis=0)
row_max = X.max(axis=1)
best_feature_per_row = X.argmax(axis=1)
```

```python
# check
assert col_means.shape == (3,) and np.allclose(col_means, X.sum(0) / 100)
assert row_max.shape == (100,) and np.all(row_max >= X[:, 0])
assert np.array_equal(X[np.arange(100), best_feature_per_row], row_max)
print("✅ Exercise 3 passed")
```

### Exercise 4: Standardization with broadcasting
Write `standardize(X)`, which returns X with every **column** rescaled to mean 0 and standard deviation 1.
No loops allowed. This is exactly what sklearn's `StandardScaler` does (Module 10).

```python
def standardize(X):
    return (X - X.mean(axis=0)) / X.std(axis=0)
```

```python
# check
Z = standardize(X)
assert Z.shape == X.shape
assert np.allclose(Z.mean(axis=0), 0), "column means should be 0"
assert np.allclose(Z.std(axis=0), 1), "column stds should be 1"
print("✅ Exercise 4 passed")
```

### Exercise 5: Row normalization
Write `normalize_rows(X)`, which divides every **row** by its own sum so that each row sums to 1.
Hint: use `keepdims`.

```python
def normalize_rows(X):
    return X / X.sum(axis=1, keepdims=True)
```

```python
# check
P = normalize_rows(np.array([[1.0, 3.0], [2.0, 2.0], [0.0, 5.0]]))
assert np.allclose(P, [[0.25, 0.75], [0.5, 0.5], [0.0, 1.0]])
print("✅ Exercise 5 passed")
```

### Exercise 6: Euclidean distances (very useful for KNN and k-means later)
Given a single point `p` of shape `(2,)` and points `pts` of shape `(n, 2)`, return an array of shape `(n,)`
holding the distance from `p` to each point: $\sqrt{(x_1 - p_1)^2 + (x_2 - p_2)^2}$.

```python
def distances(p, pts):
    return np.sqrt(((pts - p) ** 2).sum(axis=1))
```

```python
# check
d = distances(np.array([0.0, 0.0]), np.array([[3.0, 4.0], [1.0, 0.0], [0.0, 0.0]]))
assert np.allclose(d, [5.0, 1.0, 0.0])
print("✅ Exercise 6 passed")
```

### Exercise 7: pandas
Using the `sales` DataFrame below:
1. Add a column `revenue` = `units * price`.
2. `pune_big`: rows where the city is `"Pune"` **and** `units > 10`.
3. `rev_by_city`: total revenue per city as a Series, sorted from largest to smallest.

```python
sales = pd.DataFrame({
    "city":  ["Pune", "Oslo", "Pune", "Leeds", "Oslo", "Pune", "Leeds"],
    "units": [12, 5, 8, 20, 15, 30, 3],
    "price": [2.5, 10.0, 2.5, 4.0, 10.0, 2.0, 4.0],
})

sales["revenue"] = sales["units"] * sales["price"]
pune_big = sales[(sales["city"] == "Pune") & (sales["units"] > 10)]
rev_by_city = sales.groupby("city")["revenue"].sum().sort_values(ascending=False)
```

```python
# check
assert "revenue" in sales and sales["revenue"].sum() == 30 + 50 + 20 + 80 + 150 + 60 + 12
assert len(pune_big) == 2 and set(pune_big["units"]) == {12, 30}
assert list(rev_by_city.index) == ["Oslo", "Pune", "Leeds"] and rev_by_city["Oslo"] == 200
print("✅ Exercise 7 passed")
```
