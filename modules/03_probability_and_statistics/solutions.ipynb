### Exercise 1: Statistics from scratch
Implement these without using `np.mean`, `np.var`, `np.std` or `np.median` (`np.sum`, `np.sort` and `len` are fine):
- `mean(x)`
- `variance(x, ddof=0)`: divide by `n - ddof`
- `std(x, ddof=0)`
- `median(x)`: handle both odd and even lengths

```python
def mean(x):
    return np.sum(x) / len(x)

def variance(x, ddof=0):
    return np.sum((x - mean(x)) ** 2) / (len(x) - ddof)

def std(x, ddof=0):
    return np.sqrt(variance(x, ddof))

def median(x):
    s = np.sort(x)
    n = len(s)
    mid = n // 2
    return s[mid] if n % 2 == 1 else (s[mid - 1] + s[mid]) / 2
```

```python
# check
for size in [7, 10]:
    x = rng.normal(5, 2, size)
    assert np.isclose(mean(x), np.mean(x))
    assert np.isclose(variance(x), np.var(x)) and np.isclose(variance(x, 1), np.var(x, ddof=1))
    assert np.isclose(std(x, 1), np.std(x, ddof=1))
    assert np.isclose(median(x), np.median(x))
print("✅ Exercise 1 passed")
```

### Exercise 2: z-scores and outliers
Write `zscore_outliers(x, threshold=3)`, which returns a **boolean mask** marking values whose |z-score| exceeds the threshold.

```python
def zscore_outliers(x, threshold=3):
    z = (x - x.mean()) / x.std()
    return np.abs(z) > threshold
```

```python
# check
data = np.concatenate([rng.normal(0, 1, 1000), [8.0, -9.0]])
mask = zscore_outliers(data)
assert mask.dtype == bool and mask.shape == data.shape
assert mask[-1] and mask[-2], "the injected outliers should be flagged"
assert mask.sum() <= 6
print("✅ Exercise 2 passed; outliers:", data[mask].round(2))
```

### Exercise 3: Probability by simulation
Estimate by simulation (at least 100,000 trials) the probability that, when rolling **three** dice,
**at least two show the same number**. Store it in `p_match`.
(Exact answer: 1 - 6·5·4/6³ = 0.444...)

```python
rolls = rng.integers(1, 7, size=(200_000, 3))
all_different = (rolls[:, 0] != rolls[:, 1]) & (rolls[:, 0] != rolls[:, 2]) & (rolls[:, 1] != rolls[:, 2])
p_match = 1 - all_different.mean()
```

```python
# check
assert abs(p_match - (1 - 120 / 216)) < 0.01, p_match
print("✅ Exercise 3 passed ->", round(p_match, 4))
```

### Exercise 4: Bayes' rule for a spam filter
- 20% of emails are spam.
- The word "free" appears in 60% of spam emails.
- The word "free" appears in 5% of non-spam emails.

Write `posterior(prior, likelihood_pos, likelihood_neg)`, which returns P(class | evidence). Use it to compute
`p_spam_given_free`.

```python
def posterior(prior, likelihood_pos, likelihood_neg):
    evidence = likelihood_pos * prior + likelihood_neg * (1 - prior)
    return likelihood_pos * prior / evidence

p_spam_given_free = posterior(0.2, 0.6, 0.05)
```

```python
# check
assert np.isclose(p_spam_given_free, 0.12 / (0.12 + 0.04))
assert np.isclose(posterior(0.01, 0.95, 0.05), 0.0095 / (0.0095 + 0.0495))
print(f"✅ Exercise 4 passed -> P(spam | 'free') = {p_spam_given_free:.3f}")
```

### Exercise 5: Standard error
Simulate the CLT. Draw `n_experiments = 5000` samples of size `n = 50` from a **uniform(0, 10)** distribution,
compute each sample's mean, and store the standard deviation of those means in `se_simulated`.
Compute the theoretical value `se_theory = sigma / sqrt(n)`. (The std of Uniform(a, b) is $(b-a)/\sqrt{12}$.)

```python
sample_means = rng.uniform(0, 10, size=(5000, 50)).mean(axis=1)
se_simulated = sample_means.std()
se_theory = (10 / np.sqrt(12)) / np.sqrt(50)
```

```python
# check
assert np.isclose(se_theory, 0.40825, atol=1e-4)
assert abs(se_simulated - se_theory) < 0.02
print(f"✅ Exercise 5 passed -> simulated {se_simulated:.4f} vs theory {se_theory:.4f}")
```

### Exercise 6: Correlation from scratch
Implement `pearson_r(x, y)` using the covariance formula from the lesson (no `np.corrcoef`), then find which
feature in `F` is **most strongly** related to `target`, in either direction. Store its column index in `strongest`.

```python
def pearson_r(x, y):
    cov = np.mean((x - x.mean()) * (y - y.mean()))
    return cov / (x.std() * y.std())

n = 500
F = rng.normal(size=(n, 4))
target = 0.3 * F[:, 0] - 2.0 * F[:, 2] + rng.normal(size=n)

rs = np.array([pearson_r(F[:, j], target) for j in range(F.shape[1])])
strongest = int(np.abs(rs).argmax())
```

```python
# check
for j in range(4):
    assert np.isclose(pearson_r(F[:, j], target), np.corrcoef(F[:, j], target)[0, 1])
assert strongest == 2, "the strongest relationship is negative; did you use abs()?"
print("✅ Exercise 6 passed")
```
