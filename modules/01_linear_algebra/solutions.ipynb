### Exercise 1: The dot product by hand
Implement `dot(u, v)` **with a loop** (no `@` or `np.dot`). The check compares it with NumPy.

```python
def dot(u, v):
    total = 0.0
    for a, b in zip(u, v):
        total += a * b
    return total
```

```python
# check
u, v = rng.normal(size=5), rng.normal(size=5)
assert np.isclose(dot(u, v), u @ v)
print("✅ Exercise 1 passed")
```

### Exercise 2: L2 norm and distance, without `np.linalg.norm`

```python
def l2_norm(v):
    return np.sqrt(np.sum(v ** 2))

def euclidean_distance(a, b):
    return l2_norm(a - b)
```

```python
# check
assert np.isclose(l2_norm(np.array([3.0, 4.0])), 5.0)
a, b = rng.normal(size=4), rng.normal(size=4)
assert np.isclose(euclidean_distance(a, b), np.linalg.norm(a - b))
print("✅ Exercise 2 passed")
```

### Exercise 3: Cosine similarity
Return the cosine of the angle between `a` and `b`. Then use it to answer the question: which document vector
is **most similar** to `query`? Store that document's index in `best_doc`.

(Each document is a count of the words `[ml, python, cooking, recipe]`.)

```python
def cosine_similarity(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))

docs = np.array([
    [5, 3, 0, 0],   # doc 0: ML + python
    [0, 1, 6, 4],   # doc 1: cooking
    [1, 0, 2, 5],   # doc 2: recipes
], dtype=float)
query = np.array([2, 2, 0, 1], dtype=float)

sims = np.array([cosine_similarity(query, d) for d in docs])
best_doc = int(sims.argmax())
```

```python
# check
assert np.isclose(cosine_similarity(np.array([1., 0.]), np.array([0., 2.])), 0.0)
assert np.isclose(cosine_similarity(np.array([1., 1.]), np.array([3., 3.])), 1.0)
assert best_doc == 0
print("✅ Exercise 3 passed")
```

### Exercise 5: Matrix multiplication with loops
Implement `matmul(A, B)` using loops and your `dot` function from Exercise 1. Raise `ValueError` if the
shapes are incompatible. This builds intuition for "row of A · column of B".

```python
def matmul(A, B):
    n, k = A.shape
    k2, m = B.shape
    if k != k2:
        raise ValueError(f"incompatible shapes {A.shape} and {B.shape}")
    C = np.zeros((n, m))
    for i in range(n):
        for j in range(m):
            C[i, j] = dot(A[i, :], B[:, j])
    return C
```

```python
# check
A, B = rng.normal(size=(3, 4)), rng.normal(size=(4, 2))
assert np.allclose(matmul(A, B), A @ B)
try:
    matmul(A, A)
    raise AssertionError("should have raised ValueError")
except ValueError:
    pass
print("✅ Exercise 5 passed")
```

### Exercise 6: Vectorized predictions
A linear model has weights `w` and bias `b`. Compute the predictions for all rows of `X` in **one line**,
without loops.

```python
X = rng.normal(size=(8, 3))
w = np.array([0.5, -1.0, 2.0])
b = 3.0
y_pred = X @ w + b
```

```python
# check
assert y_pred.shape == (8,)
assert np.allclose(y_pred, [sum(X[i, j] * w[j] for j in range(3)) + b for i in range(8)])
print("✅ Exercise 6 passed")
```

### Exercise 7: Solve a linear system
A shop sells apples and bananas.
- Order 1: 3 apples + 2 bananas cost 13.
- Order 2: 1 apple + 4 bananas cost 11.

Find the price of each fruit using `np.linalg.solve`. Store the result in `prices = [apple, banana]`.

```python
A = np.array([[3.0, 2.0],
              [1.0, 4.0]])
y = np.array([13.0, 11.0])
prices = np.linalg.solve(A, y)
```

```python
# check
assert np.allclose(prices, [3.0, 2.0])
print("✅ Exercise 7 passed")
```

### Exercise 8: Least-squares fit with an intercept
The data below follows `y = 1.5*x1 - 2*x2 + 7 + noise`.
Build a design matrix with a column of ones, then use `np.linalg.lstsq` to recover
`w_hat = [intercept, w1, w2]`.

```python
n = 200
x1, x2 = rng.uniform(-5, 5, n), rng.uniform(-5, 5, n)
y = 1.5 * x1 - 2.0 * x2 + 7.0 + rng.normal(0, 0.5, n)

X_design = np.column_stack([np.ones(n), x1, x2])
w_hat, *_ = np.linalg.lstsq(X_design, y, rcond=None)
```

```python
# check
assert X_design.shape == (200, 3)
assert np.allclose(w_hat, [7.0, 1.5, -2.0], atol=0.2), w_hat
print("✅ Exercise 8 passed  ->", w_hat)
```
